```python
from underthesea import pos_tag
import sys
sys.path.append('/home/vlai-vqa-nle/minhtq/vqa-nle/ms-swift/examples/train/grpo/plugin')
from reasoning_rewards import ReasoningRewardScorer

# Khởi tạo scorer
scorer = ReasoningRewardScorer(threshold=0.3)

# Sample 1: Câu hỏi về màu sắc (what color - cluster 2)
completion1 = "<REASONING>Trong ảnh có một chiếc xe màu đỏ</REASONING><ANSWER>đỏ</ANSWER>"
solution1 = "<questiontype>what color</questiontype><explain>Xe trong ảnh có màu đỏ</explain><answer>đỏ</answer>"

# Sample 2: Câu hỏi về trạng thái (is the - cluster 1)
completion2 = "<REASONING>Người đàn ông đang chơi tennis trên sân</REASONING><ANSWER>có</ANSWER>"
solution2 = "<questiontype>is the</questiontype><explain>Anh ấy đang chơi tennis</explain><answer>có</answer>"

# Sample 3: Câu hỏi mở (what - cluster 6)
completion3 = "<REASONING>Đây là một chiếc bánh pizza với nhiều topping</REASONING><ANSWER>pizza</ANSWER>"
solution3 = "<questiontype>what</questiontype><explain>Món ăn là pizza có phô mai và rau</explain><answer>pizza</answer>"

# Tính reward cho từng sample
samples = [
    (completion1, solution1, "Sample 1 - What color"),
    (completion2, solution2, "Sample 2 - Is the"),
    (completion3, solution3, "Sample 3 - What")
]

print("=== KẾT QUẢ REWARD ===\n")
for comp, sol, name in samples:
    reward = scorer.reasoning_reward(comp, sol)
    print(f"{name}: Reward = {reward:.4f}")
    print()

# Test batch
print("\n=== TEST BATCH ===")
completions = [completion1, completion2, completion3]
solutions = [solution1, solution2, solution3]
rewards = scorer.reasoning_rewards_batch(completions, solutions)
print(f"\nRewards batch: {[f'{r:.4f}' for r in rewards]}")
```
